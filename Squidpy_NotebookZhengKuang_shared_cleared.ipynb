# 1. Overview of Squidpy

## 1.1 Spatial transcriptomics

Spatial multi-omics provides unprecedented means for resolving tissue architecture and cell-cell communication, and the resulting data volumes are growing rapidly. This motivates a tool capable of flexible storage, integration, and visualisation of spatial multi-omics data (such as spatial transcriptomics and spatial multiplexed proteomics). Squidpy, the subject of this tutorial, is such a tool. It integrates a range of methods and algorithms into a one-stop framework for controlling and interactively visualising spatial multi-omics data. The original Squidpy paper was published in *Nature Methods*: "Squidpy: a scalable framework for spatial omics analysis" [1].

$Figure 1$



## 1.2 Background

Single-cell RNA-seq (scRNA-seq) necessarily loses the spatial coordinates of cells during tissue dissociation. Spatial transcriptomics fills this gap by providing in situ information at cellular, and even subcellular, resolution. Common spatial-omics modalities include spatial transcriptomics, spatial proteomics, and spatial metabolomics; these have distinct data formats and analysis methods, which poses a challenge for analysis software. The algorithms involved are also diverse, spanning spatial clustering (e.g. BayesSpace) and deep-learning-based image segmentation. To address these challenges, Squidpy (Spatial Quantification of Molecular Data in Python) was developed (Figure 1). Its core goal is to integrate diverse spatial-omics data and present them through a unified data format, analysis, and interactive visualisation. Squidpy supports exploration of data in three dimensions to better resolve spatial patterns and morphology (Figure 1c). Squidpy is built on scanpy (so its primary data object is AnnData) together with libraries such as scikit-image, napari, and Dask. The official documentation is at https://squidpy.readthedocs.io/en/stable/notebooks/tutorials/index.html.

## 1.3 Key results

### 1.3.1 Squidpy provides tools for identifying spatial patterns in tissue

Identifying spatial neighbours relies on a spatial graph. For example, spatial transcriptomics technologies (Visium, DBit-seq, etc.) record transcript spatial information using square or hexagonal arrays of spot coordinates (Figure 2a); other image-based technologies such as seqFISH, MERFISH, Imaging Mass Cytometry, CyCIF, 4i, and Spatial Metabolomics follow the same principle. Euclidean distance and Delaunay triangulation can be used to construct the neighbourhood graph. Using the latter, the authors performed permutation-based neighbourhood enrichment across cell-type annotations on a recently published seqFISH dataset, recovering cell types consistent with the original publication (Figure 2b); their spatial neighbourhood relationships are shown in Figure 2c. In 3D spatial coordinates, Endothelial 2 and Pericytes showed strong neighbourhood enrichment (Figure 2d), while Ependymal differed from all other cell types (Figure 2e). Compared with Giotto, which performs comparable tasks, these computations took roughly one tenth of the time (Figure S1a,b). Given that data such as 4i can provide subcellular resolution, the authors tested a dataset of about 270,000 pixels (Figure 2f) and computed the co-occurrence of clusters as a function of distance (Figure 2g), showing strong within-cluster co-occurrence. The same phenomenon was observed in a separate Slide-seqV2 dataset (Figure 2i). Using Moran's I spatial autocorrelation statistics, the authors identified the three genes with the strongest spatially specific expression: *Ttr*, *Mbp*, and *Hpca* (Figure 2k).

### 1.3.2 Squidpy can analyse large image (field-of-view) data

To improve image analysis, the authors introduced a new data container, ImageContainer (Figure 3a). It efficiently stores image data, supports flexible transfer between disk and memory, and integrates seamlessly with AnnData, napari, Dask Image, and scikit-image, providing a better way to couple image information with molecular expression maps and enabling convenient cropping of images together with their expression matrices. Image processing can be loaded and processed with multi-threading (Figure S1b), and merging multiple images corrects for boundary effects automatically. Squidpy can integrate Cellpose and StarDist algorithms at the level of nuclei, cells, counts, field-of-view size, and segmentation masks (Figure S2b). For example, in cell segmentation, DAPI-stained fluorescence was compared against 10X Visium spatial transcriptomics data of the mouse hippocampus, showing high consistency (Figure 3b-d). For data where spot resolution exceeds cell size, Squidpy also supports deconvolution via Tangram or Cell2location.

### 1.3.3 Squidpy enables integrated analysis of spatial transcriptomics data

Squidpy integrates many functions; among the most commonly used is the identification of spatially variable genes through Moran's I spatial correlation statistics. For example, *Mobp* and *Nrgn* were defined as spatially specific genes in mouse brain sections (Figure 4a,b), and these genes also coincide with particular cell types (Figure 4d). For cell-cell communication, Squidpy provides a reimplementation of CellPhoneDB that runs faster than Giotto and even CellPhoneDB itself (Figure S1b,d). In addition, Squidpy supports both gene-based clustering (Figure 4d) and image-based clustering (Figure 4h); the high reproducibility between the two demonstrates consistency between tissue morphology and molecular function. Differences between the two clustering approaches also exist (Figure 4i): image-based clustering offers an alternative perspective for tissues with clear structural differences such as vasculature, tissue boundaries, and fibrotic regions.

## 3.2 Spatial co-occurrence

Using the raw spatial coordinates, the co-occurrence of clusters can be quantified. The co-occurrence score is defined as $\frac{p(exp|cond)}{p(exp)}$, where $p(exp|cond)$ is the conditional probability of observing a cluster given a conditioning cluster, and $p(exp)$ is the probability of observing the cluster within the target radius. The score is computed over increasing radii around each cell. Computation is performed by `squidpy.gr.co_occurrence()` and visualisation by `squidpy.pl.co_occurrence()`.


```python
sq.gr.co_occurrence(adata, cluster_key="cell type")
sq.pl.co_occurrence(
    adata,
    cluster_key="cell type",
    clusters=["basal CK tumor cell", "T cells"],
    figsize=(15, 4),
)
# T cells co-occur with endothelial and vimentin-high stromal cells
# basal CK tumour cells appear self-aggregated, closest to small elongated stromal cells
```

## 3.3 `Neighborhood enrichment`

Similar to the co-occurrence analysis in 3.2, `neighborhood enrichment` reveals spatial proximity relationships between clusters. It is computed by `squidpy.gr.nhood_enrichment()`, a permutation-based test; `n_perms` controls the number of permutations (default 1000). The resulting enrichment score reflects spatial proximity: pairs of points from two clusters that lie close together receive a high score, and vice versa.


```python
# compute the connectivity matrix
sq.gr.spatial_neighbors(adata)

# compute neighborhood enrichment
sq.gr.nhood_enrichment(adata, cluster_key="cell type")

# visualise
sq.pl.nhood_enrichment(adata, cluster_key="cell type")
# T cells again show a high enrichment score with stromal and endothelial cells
# apoptotic tumour cells are evenly distributed throughout the tissue
```

## 3.4 Interaction matrix and network centrality

Squidpy provides descriptive statistics for spatial graphs. The interaction matrix counts the number of edges shared between each pair of clusters, computable via `squidpy.gr.interaction_matrix()` and visualisable via `squidpy.pl.interaction_matrix()`.


```python
sq.gr.interaction_matrix(adata, cluster_key="cell type")
sq.pl.interaction_matrix(adata, cluster_key="cell type")
# interactions among clusters are generally low, except for apoptotic tumour cells
```

Network centralities such as `degree_centrality`, `average_clustering`, and `closeness_centrality` can be computed with `squidpy.gr.centrality_scores()` and visualised with `squidpy.pl.centrality_scores()`.

```python
sq.gr.centrality_scores(
    adata,
    cluster_key="cell type",
)
sq.pl.centrality_scores(adata, cluster_key="cell type", figsize=(20, 5), s=500)
# apoptotic tumour cells show tight (high) centrality
```

## 5.3 `Ripley's` statistics

`Ripley's` statistics describe whether a discrete annotation (e.g. cell type or cluster) is clustered, dispersed, or randomly distributed within a region. Computation is via `squidpy.gr.ripley()` and visualisation via `squidpy.pl.ripley()`. See https://squidpy.readthedocs.io/en/stable/notebooks/examples/graph/compute_ripley.html for algorithmic details.


```python
mode = "L"
sq.gr.ripley(adata, cluster_key="cluster", mode=mode, max_dist=500)
sq.pl.ripley(adata, cluster_key="cluster", mode=mode)
# higher values indicate more clustered distributions (e.g. CA1_CA2_CA3_Subiculum cells)
# lower values indicate more dispersed distributions (e.g. Mural cells)
```

## 6.4 Clustering with image features

This module comprises `summary`, `histogram`, and `texture` features; see:
- Extract summary features: https://squidpy.readthedocs.io/en/stable/notebooks/examples/image/compute_summary_features.html
- Extract histogram features: https://squidpy.readthedocs.io/en/stable/notebooks/examples/image/compute_histogram_features.html
- Extract texture features: https://squidpy.readthedocs.io/en/stable/notebooks/examples/image/compute_texture_features.html


```python
# define the parameters
params = {
    "features_orig": {
        "features": ["summary", "texture", "histogram"],
        "scale": 1.0,
        "mask_circle": True,
    },
    # compute summary and histogram features at native resolution
    "features_context": {"features": ["summary", "histogram"], "scale": 1.0},
    # compute summary and histogram features at lower resolution
    "features_lowres": {"features": ["summary", "histogram"], "scale": 0.25},
}

for feature_name, cur_params in params.items():
    # results are stored in `adata.obsm[feature_name]`
    sq.im.calculate_image_features(
        adata, img, layer="image", key_added=feature_name, n_jobs=1, **cur_params
    )

# combine the results into a single dataframe
adata.obsm["features"] = pd.concat(
    [adata.obsm[f] for f in params.keys()], axis="columns"
)

# ensure feature names are unique
adata.obsm["features"].columns = ad.utils.make_index_unique(
    adata.obsm["features"].columns
)
```

The image features computed above can be used to re-cluster the data; a helper function that clusters features is defined below:

```python
def cluster_features(features: pd.DataFrame, like=None):
    """
    Calculate leiden clustering of features.

    Specify filter of features using `like`.
    """
    # filter features
    if like is not None:
        features = features.filter(like=like)
    # create a temporary AnnData for clustering
    adata = ad.AnnData(features)
    # important: features are unscaled, and must be scaled before PCA
    sc.pp.scale(adata)
    # compute leiden clustering
    sc.pp.pca(adata, n_comps=min(10, features.shape[1] - 1))
    sc.pp.neighbors(adata)
    sc.tl.leiden(adata)

    return adata.obs["leiden"]
```

```python
# cluster each type of feature and visualise
adata.obs["features_summary_cluster"] = cluster_features(
    adata.obsm["features"], like="summary"
)
adata.obs["features_histogram_cluster"] = cluster_features(
    adata.obsm["features"], like="histogram"
)
adata.obs["features_texture_cluster"] = cluster_features(
    adata.obsm["features"], like="texture"
)

sc.set_figure_params(facecolor="white", figsize=(8, 8))
sq.pl.spatial_scatter(
    adata,
    color=[
        "features_summary_cluster",
        "features_histogram_cluster",
        "features_texture_cluster",
        "cluster",
    ],
    ncols=3,
)
# the clustering of different feature types differs, and is more dispersed than the original clusters
```

## 7.2 `Image features`

The new image features can be used for clustering, so that morphological features are taken into account during annotation:

```python
# define a function that clusters image features
def cluster_features(features: pd.DataFrame, like=None) -> pd.Series:
    """
    Calculate leiden clustering of features.

    Specify filter of features using `like`.
    """
    # filter features
    if like is not None:
        features = features.filter(like=like)
    # build a temporary AnnData for clustering
    adata = ad.AnnData(features)
    # features are unscaled, so scale before PCA
    sc.pp.scale(adata)
    # compute leiden clustering
    sc.pp.pca(adata, n_comps=min(10, features.shape[1] - 1))
    sc.pp.neighbors(adata)
    sc.tl.leiden(adata)

    return adata.obs["leiden"]


# cluster using the features (here only summary features are present)
adata.obs["features_cluster"] = cluster_features(adata.obsm["features"],
                                                 like="summary")

# compare the feature-based clusters with the original clusters
sq.pl.spatial_scatter(adata, color=["features_cluster", "cluster"])
# e.g. the regions around Fiber_tract and Hippocampus are consistent between
# the two clusterings, while the Cortex differs markedly; overall the
# feature-based clustering is more dispersed
```

## 3.3 `Neighborhood enrichment`

Similar to the co-occurrence analysis in 3.2, `neighborhood enrichment` reveals spatial proximity relationships between clusters. It is computed by `squidpy.gr.nhood_enrichment()`, a permutation-based test; `n_perms` controls the number of permutations (default 1000). The resulting enrichment score reflects spatial proximity: pairs of points from two clusters that lie close together receive a high score, and vice versa.

## 3.2 Spatial co-occurrence

Using the raw spatial coordinates, the co-occurrence of clusters can be quantified. The co-occurrence score is defined as $\frac{p(exp|cond)}{p(exp)}$, where $p(exp|cond)$ is the conditional probability of observing a cluster given a conditioning cluster, and $p(exp)$ is the probability of observing the cluster within the target radius. The score is computed over increasing radii around each cell. Computation is performed by `squidpy.gr.co_occurrence()` and visualisation by `squidpy.pl.co_occurrence()`.


```python
sq.gr.co_occurrence(adata, cluster_key="cluster")
sq.pl.co_occurrence(
    adata,
    cluster_key="cluster",
    clusters="Hippocampus",
    figsize=(8, 4),
)
# consistent with the earlier results, Pyramidal_layer co-occurs at short distances
# while Hippocampus contributes over larger distances; these results depend on
# the abundance and spatial distribution of each cluster
```

## 9.3 Extracting mean cell intensity

For the three channels/markers above, the segmentation results allow the mean intensity per cell to be computed.
### 9.3.1 Converting the data to CMYK
Before computation, the colour space is converted from `RGB` to `CMYK` using the provided `rgb2cmyk` function:


```python
def rgb2cmyk(arr):
    """Convert arr from RGB to CMYK color space."""
    R = arr[..., 0] / 255
    G = arr[..., 1] / 255
    B = arr[..., 2] / 255
    K = 1 - (np.max(arr, axis=-1) / 255)
    C = (1 - R - K) / (1 - K + np.finfo(float).eps)  # avoid division by 0
    M = (1 - G - K) / (1 - K + np.finfo(float).eps)
    Y = (1 - B - K) / (1 - K + np.finfo(float).eps)
    return np.stack([C, M, Y, K], axis=3)


img.apply(rgb2cmyk, layer="image", new_layer="image_cmyk", copy=False)
img.show("image_cmyk", channelwise=True)
```

### 9.3.2 Computing mean intensity

A custom function `segmentation_image_intensity` is provided to compute and return a new `layer`:


```python
def segmentation_image_intensity(arr, image_cmyk):
    """
    Calculate per-channel mean intensity of the center segment.

    arr: the segmentation
    image_cmyk: the raw image values
    """
    import skimage.measure

    # the center of the segmentation mask contains the current label
    # use that to calculate the mask
    s = arr.shape[0]
    mask = (arr == arr[s // 2, s // 2, 0, 0]).astype(int)

    # ensure skimage is installed
    # use skimage.measure.regionprops to get the intensity per channel
    features = []
    for c in range(image_cmyk.shape[-1]):
        feature = skimage.measure.regionprops_table(
            np.squeeze(mask),  # skimage needs 3d or 2d images, so squeeze excess dims
            intensity_image=np.squeeze(image_cmyk[:, :, :, c]),
            properties=["mean_intensity"],
        )["mean_intensity"][0]
        features.append(feature)
    return features
```

```python
# compute the features
sq.im.calculate_image_features(
    adata,
    img,
    library_id="library_id",
    features="custom",
    spot_scale=10,
    layer="segmentation",
    features_kwargs={
        "custom": {
            "func": segmentation_image_intensity,
            "additional_layers": ["image_cmyk"],
        }
    },
)
```

Results are stored in `adata.obsm['img_features']`; channel 0 corresponds to `145_CD45`, channel 1 to `174_CK`, and channel 2 to `113_vimentin`.

```python
# compare the transformed values with the intensities computed above
channels = ["CD45", "CK", "vimentin"]

fig, axes = plt.subplots(1, 3, figsize=(15, 3))
for i, ax in enumerate(axes):
    X = np.array(adata[:, channels[i]].X.todense())[:, 0]
    Y = adata.obsm["img_features_transformed"][f"segmentation_image_intensity_{i}"]
    ax.scatter(X, Y)
    ax.set_xlabel("true value in adata.X")
    ax.set_ylabel("computed mean intensity")
    corr = np.corrcoef(X, Y)[1, 0]
    ax.set_title(f"{channels[i]}, corr: {corr:.2f}")
```

It is computed by `squidpy.gr.nhood_enrichment()`, a permutation-based test; `n_perms` controls the number of permutations (default 1000). The resulting enrichment score reflects spatial proximity: pairs of points from two clusters that lie close together receive a high score, and vice versa.


```python
# compute the connectivity matrix
sq.gr.spatial_neighbors(adata, coord_type="generic")
# neighborhood enrichment
sq.gr.nhood_enrichment(adata, cluster_key="cluster")
# visualise
sq.pl.nhood_enrichment(adata, cluster_key="cluster",
                       method="ward", vmin=-100, vmax=100)
```

## 3.2 Spatial co-occurrence

Using the raw spatial coordinates, the co-occurrence of clusters can be quantified. ### 12.2.2 Computing mean expression

Compute the mean expression per `Leiden cluster` of the query dataset:

```python
ser_counts = adata.obs["leiden"].value_counts()
ser_counts.name = "cell counts"
meta_leiden = pd.DataFrame(ser_counts)

cat_name = "leiden"
sig_leiden = pd.DataFrame(
    columns=adata.var_names, index=adata.obs[cat_name].cat.categories
)
for clust in adata.obs[cat_name].cat.categories:
    sig_leiden.loc[clust] = adata[adata.obs[cat_name].isin([clust]), :].X.mean(0)
sig_leiden = sig_leiden.transpose()
leiden_clusters = ["Leiden-" + str(x) for x in sig_leiden.columns.tolist()]
sig_leiden.columns = leiden_clusters
meta_leiden.index = sig_leiden.columns.tolist()
meta_leiden["leiden"] = pd.Series(
    meta_leiden.index.tolist(), index=meta_leiden.index.tolist()
)
```

### 12.2.3 Assigning cell types by highly expressed markers

The number of clusters is much larger than the number of cell types, so annotation can proceed with confidence:
